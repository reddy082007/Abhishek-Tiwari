```python
print("Hello! I am Python.")
print("Today we will build a cleaning robot together. 🤖")

print("9 + 2 =", 9 + 2)

# D = Dirty, C = Clean
room = ["D", "C", "D", "D", "C"]


def show_room(room):
    picture = ""
    for spot in room:
        if spot == "D":
            picture += "💩"
        else:
            picture += "✨"
    print(picture)


print("\nOur room right now:")
show_room(room)


def clean_spot(spot):
    # If dirty, clean it.
    return "C"


result = clean_spot("D")
print("\nRobot looked at a dirty spot 💩 and made it:", result, "(C means Clean ✨)")

print("\nBefore cleaning:")
show_room(room)

# Clean each spot one by one
for i in range(len(room)):
    room[i] = clean_spot(room[i])
    print(f"After cleaning spot {i + 1}:")
    show_room(room)

print("\nAfter - All done! ✨🤖✨")
show_room(room)
```
